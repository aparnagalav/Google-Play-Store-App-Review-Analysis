### Data Wrangling Code

````python
# 1 Removing duplicate rows
playstore_df = playstore_df.drop_duplicates()
reviews_df = reviews_df.drop_duplicates()
````

````python
# drops rows or columns that contain missing values.
playstore_df = playstore_df.dropna(subset=["Type", "Content Rating", "Current Ver", "Android Ver"])
````

````python
reviews_df = reviews_df.dropna()
````

````python
# verifying the null values
playstore_df.isnull().sum()
````

````python
reviews_df.isnull().sum()
````

````python
playstore_df.info() # here Reviews are in object but the values are numeric (159, 967, 4500, 100000)
# Installs are in object but the values are in numeric (1000+, 10000+) same goes for Price, Size
````

````python
reviews_df.info()
````

````python
# coverting the data in numeric values for Reviews column
playstore_df["Reviews"] = pd.to_numeric(playstore_df["Reviews"])
````

````python
#verifying the details
playstore_df["Reviews"].dtype
````

````python
playstore_df["Installs"].dtype
````

````python
# replacing the comma in Installs
playstore_df["Installs"] = playstore_df["Installs"].str.replace(",","")
````

````python
## replacing the "+" in Installs #Regex(regular expression) = false
playstore_df["Installs"] = playstore_df["Installs"].str.replace("+", "", regex=False)
````

````python
# converting into numeric values
playstore_df["Installs"] = pd.to_numeric(playstore_df["Installs"])
````

````python
#verifying
playstore_df["Installs"].dtype
````

````python
## replacing the $ in Price #Regex(regular expression) = false
playstore_df["Price"] = playstore_df["Price"].str.replace("$", "", regex=False)
````

````python
# converting into numeric values
playstore_df["Price"] = pd.to_numeric(playstore_df["Price"])
````

````python
#verifying
playstore_df["Price"].dtype
````

````python
#Inspecting the data
playstore_df["Size"].unique()
````

````python
playstore_df["Size"].value_counts()
````

````python
#Replace "Varies with device" with NaN
playstore_df["Size"] = playstore_df["Size"].replace("Varies with device", np.nan)
````

````python
playstore_df["Size"] = playstore_df["Size"].apply(lambda x: float(x[:-1]) / 1024 if isinstance(x, str) and x.endswith("k") else x)
````

````python
#removing the "M" by replacing
playstore_df["Size"] = playstore_df["Size"].str.replace("M", "", regex=False)
````

````python
playstore_df["Size"] = pd.to_numeric(playstore_df["Size"], errors="coerce")
````
